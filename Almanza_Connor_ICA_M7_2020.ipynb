<h1>While Loops</h1>
<ul style="font-size:20px;">
   <li>4.1 The while Loop</li>
    <li>4.3 Application: Processing Sentinel Values</li>
    <li>4.5 Common Loop Algorithms</li>
</ul>
<ul>
<li>while loop - Tells the computer to execute a set of instructions while a specified condition is True.</li>
<li>In other words, while loops keep executing the same group of instructions until the condition becomes False.</li>
<li>It often happens that you want to execute a sequence of statements a given number of times. You can use a while loop that is controlled by a counter</li>
<li>This is known as a <b>count-controlled</b> or <b>definite</b> loop. The loop body will be executed a definite number of times.</li>
</ul>

```python
#Example Convert the following to a while loop:
balance = 2000
monthly_deposit = 100
interest_rate = .00416
number_of_periods = 3
'''
1. For investment accounts (like savings, CD's, or brokerage accounts), 
   financial institutions typically calculate interest before adding new deposits or withdrawing funds for a given compounding period.
2. The fees should generally be applied after the period's compound interest calculation.
'''
print(f"Beginning Balance: {balance:.2f}")
balance += (balance * interest_rate) + monthly_deposit #period 1
print(f"Period 1 Balance: {balance:.2f}")
balance += (balance * interest_rate) + monthly_deposit #period 2
print(f"Period 2 Balance: {balance:.2f}")
balance += (balance * interest_rate) + monthly_deposit #period 3
print(f"Period 3 Balance: {balance:.2f}")

#Convert the above statements to a while loop
balance = 2000
count = 0
while count < number_of_periods: #The condition is count < number_of_periods (while condition or loop header)
    balance += (balance * interest_rate) + monthly_deposit
    count += 1 # add 1 to the counter
print(f"While loop Balance: {balance:.2f}")    

```
